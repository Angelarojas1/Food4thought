#### Get flavor profile for 934 ingredients

```python
import urllib
import json
```

```python
num = 0
for i in range(2000):
    # read json files for each ingredient
    url = "https://cosylab.iiitd.edu.in/flavordb/entities_json?id={}".format(i)
    try:
        response = urllib.request.urlopen(url) 
        data = json.loads(response.read())
        
        num += 1
    
        # save json files to dropbox folder
        file ="/Users/xixi/Dropbox/food4thought/data/Flavor profile/{}.json".format(data['entity_id'])
        with open(file, 'w') as f: 
            json.dump(data, f)
    except:
        pass
```
